2.  **Upload `kaggle.json` to Colab:**

    *   Click the 'Upload to session storage' icon (looks like a page with an up arrow).

3.  **Move `kaggle.json` to the correct directory and set permissions:**

```python
def predict_and_plot(inputs, targets, name=''):
  preds=model.predict(inputs)

  accuracy= accuracy_score(targets, preds)
  print('Accuracy: {:.2f}%'.format(accuracy*100))
  cf=confusion_matrix(targets, preds, normalize='true')
  plt.figure()
  sns.heatmap(cf, annot=True)
  plt.xlabel('Prediction')
  plt.ylabel('Target')
  plt.title('{} Confusion Matrix'.format(name));
  return preds
```

```python
def random_guess(inputs):
  return np.random.choice(["No","Yes"], len(inputs))
```

```python
def all_no(inputs):
  return np.full(len(inputs), "No")
```

```python
accuracy_score(train_target, random_guess(train_target))
```

```python
accuracy_score(train_target, all_no(train_target))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize the Random Forest model with performance-tuned parameters
rf_model = RandomForestClassifier(n_estimators=100, max_depth=16, random_state=42, n_jobs=-1)

# Fit the model on our training data (using numeric and encoded columns)
rf_model.fit(x_train, train_target.values.ravel())

# Evaluate the new model
train_acc = rf_model.score(x_train, train_target) * 100
val_acc = rf_model.score(x_val, val_target) * 100
test_acc = rf_model.score(x_test, test_target) * 100

print(f"Optimized Random Forest Training Accuracy: {train_acc:.2f}%")
print(f"Optimized Random Forest Validation Accuracy: {val_acc:.2f}%")
print(f"Optimized Random Forest Test Accuracy: {test_acc:.2f}%")
```

```python
def predict_and_plot_rf(model, inputs, targets, name=''):
    preds = model.predict(inputs)
    accuracy = accuracy_score(targets, preds)
    print(f'{name} Accuracy: {accuracy*100:.2f}%')
    cf = confusion_matrix(targets, preds, normalize='true')
    plt.figure(figsize=(6, 5))
    sns.heatmap(cf, annot=True, cmap='Blues', fmt='.2f')
    plt.xlabel('Prediction')
    plt.ylabel('Target')
    plt.title(f'{name} Confusion Matrix (Random Forest)')
    plt.show()
    return preds

# Plot confusion matrices for the optimized Random Forest Model
rf_train_preds = predict_and_plot_rf(rf_model, x_train, train_target, 'Training')
rf_val_preds = predict_and_plot_rf(rf_model, x_val, val_target, 'Validation')
rf_test_preds = predict_and_plot_rf(rf_model, x_test, test_target, 'Test')
```
